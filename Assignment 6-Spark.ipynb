```python
# Then create a new SparkSession
spark = SparkSession.builder \
    .appName("Amazon Sales Analysis") \
    .master("local[*]") \
    .getOrCreate()

# Load the dataset
amazon_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("AmazonSales.csv")

# Register the DataFrame as a Temporary View
amazon_df.createOrReplaceTempView("AmazonSales")
```

# SQL Tasks: Write Spark SQL queries to answer the following questions

1. Top 5 Product Categories by Number of Orders

* Write a query to find the 5 categories with the highest number of orders.
* Output: category, number_of_orders

```python
Top_5_Categories = spark.sql("""
    SELECT category, COUNT(*) AS number_of_orders
    FROM AmazonSales
    GROUP BY category
    ORDER BY number_of_orders DESC
    LIMIT 5
""")

Top_5_Categories.show()
```

2. Unique Orders by Year

* Write a query to count the number of unique order_ids for each year.
* Hint: Use RIGHT(order_date, 4) to extract the year.
* Output: year, unique_orders

```python
Yearly_Unique_Orders = spark.sql("""
    SELECT RIGHT(order_date, 4) AS year,
    COUNT(DISTINCT order_id) AS unique_orders
  FROM
    AmazonSales
  GROUP BY
   RIGHT(order_date, 4)
  ORDER BY
    year
""")

Yearly_Unique_Orders.show()
```

3. Number of Orders by Payment Method

* Write a query to count how many orders were placed using each payment method.
* Output: payment_method, number_of_orders

```python
Payment_Method_Orders = spark.sql("""
    SELECT
      payment_method,
      COUNT(*) AS number_of_orders
    FROM
      AmazonSales
    GROUP BY
      payment_method
    ORDER BY
      number_of_orders DESC
""")

Payment_Method_Orders.show()
```

4. Average Order Total by Category

* Write a query to calculate the average total for orders in each category.
* Output: category, average_order_total

```python
Average_Category = spark.sql("""
    SELECT
      category,
      ROUND(AVG(total),2) AS average_order_total
    FROM
      AmazonSales
    GROUP BY
      category
    ORDER BY
      average_order_total DESC
""")

Average_Category.show()
```

5. Highest-Total Canceled Orders

*Write a query to display the top 5 orders with the highest total where the status is 'canceled'.
*Output: order_id, category, total, full_name

```python
Top_5_Canceled_Orders = spark.sql("""
    SELECT
      order_id,
      category,
      total,
      full_name
    FROM
      AmazonSales
    WHERE
      status = 'canceled'
    ORDER BY
      total DESC
    LIMIT 5
""")

Top_5_Canceled_Orders.show()
```
